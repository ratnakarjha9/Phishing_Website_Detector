```python
from sklearn.model_selection import GridSearchCV
```

```python
rf = RandomForestClassifier(
    random_state=42
)
```

```python
param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [None, 10, 20],
    'min_samples_split': [2, 5],
    'min_samples_leaf': [1, 2]
}
```

```python
grid_search = GridSearchCV(
    estimator=rf,
    param_grid=param_grid,
    cv=5,
    scoring='f1',
    n_jobs=-1
)
```

```python
grid_search.fit(X_train, y_train)
```

```python
print("Best Parameters:")
print(grid_search.best_params_)
```

```python
print("Best Cross-Validation F1-Score:")
print(grid_search.best_score_)
```

```python
best_rf = grid_search.best_estimator_
```

```python
y_pred_best_rf = best_rf.predict(X_test)
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred_best_rf))
print("Precision:", precision_score(y_test, y_pred_best_rf, pos_label=1))
print("Recall:", recall_score(y_test, y_pred_best_rf, pos_label=1))
print("F1-Score:", f1_score(y_test, y_pred_best_rf, pos_label=1))
```

```python
feature_importance = pd.Series(
    best_rf.feature_importances_,
    index=X.columns
)

feature_importance = feature_importance.sort_values(
    ascending=False
)

print(feature_importance)
```

```python
plt.figure(figsize=(10, 8))

feature_importance.head(15).plot(kind='barh')

plt.title("Top 15 Important Features - Random Forest")
plt.xlabel("Importance")
plt.ylabel("Feature")
plt.gca().invert_yaxis()

plt.show()
```

```python
from sklearn.metrics import classification_report

y_pred_final = best_rf.predict(X_test)

print(classification_report(
    y_test,
    y_pred_final,
    target_names=["Legitimate", "Phishing"]
))
```

```python
from sklearn.metrics import roc_auc_score

y_probability = best_rf.predict_proba(X_test)[:, 1]

auc_score = roc_auc_score(
    y_test,
    y_probability
)

print("ROC-AUC Score:", auc_score)
```

```python
from sklearn.metrics import roc_curve

fpr, tpr, thresholds = roc_curve(
    y_test,
    y_probability,
    pos_label=1
)

plt.figure(figsize=(8, 6))

plt.plot(fpr, tpr, label=f"Random Forest (AUC = {auc_score:.3f})")
plt.plot([0, 1], [0, 1], linestyle='--')

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - Random Forest")
plt.legend()
plt.show()
```
